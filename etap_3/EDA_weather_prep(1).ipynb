```python
from datetime import date
import meteostat as ms

point = ms.Point(49.39, 20.87)
start = date(2019, 6, 1)
end = date(2025, 9, 30)

stations = ms.stations.nearby(point, limit=10)
print(stations)

ts = ms.daily(stations, start, end)
df = ms.interpolate(ts, point).fetch()

print(df.shape)
print(df.head())
```

```python
df_jul_sep = df[df.index.month.isin([6, 7, 8, 9])]

df_jul_sep.head()
```

```python
df_jul_sep.tail()
```

```python
df_jul_sep['year'] = df_jul_sep.index.year
df_jul_sep['month'] = df_jul_sep.index.month

```

```python
df_jul_sep.head()
```

```python
groups = df_jul_sep.groupby(df_jul_sep.index.year)

for year, group in groups:
    print(year, group.shape)
```

```python
df_jul_sep["day"] = df_jul_sep.index.day
df_jul_sep["week"] = df_jul_sep.index.isocalendar().week
```

```python
df_jul_sep.head(20)
```

```python
col = ['temp', 'tmin', 'tmax', 'rhum', 'wspd', 'wpgt', 'pres']

mean_week = df_jul_sep.groupby(["year", "week"])[col].mean()
```
