```python
import xarray as xr

ds_instant = xr.open_dataset('./a410b01a0b3d4ee1b01b661ca754a713/data_stream-oper_stepType-instant.nc')
ds_accum = xr.open_dataset("./a410b01a0b3d4ee1b01b661ca754a713/data_stream-oper_stepType-accum.nc")
```

```python
from pathlib import Path
import xarray as xr
import pandas as pd
base_dir = Path(".")  

all_nc = list(base_dir.rglob("*.nc"))

rows = []

for file in all_nc:
    ds = xr.open_dataset(file)

    time_name = "valid_time" if "valid_time" in ds.coords else "time"
    times = pd.to_datetime(ds[time_name].values)

    rows.append({
        "file": str(file),
        "type": "accum" if "accum" in file.name else "instant" if "instant" in file.name else "other",
        "start": times.min(),
        "end": times.max(),
        "years": sorted(set(times.year)),
        "months": sorted(set(times.month)),
        "n_times": len(times)
    })

files_info = pd.DataFrame(rows)
files_info
```

```python
from pathlib import Path
import xarray as xr

base_dir = Path(".")
all_nc = sorted(base_dir.rglob("*.nc"))

instant_files = [str(f) for f in all_nc if "instant" in f.name]
accum_files = [str(f) for f in all_nc if "accum" in f.name]

print("instant:", len(instant_files))
print("accum:", len(accum_files))
print(instant_files[:3])
print(accum_files[:3])
```

```python
ds_inst = xr.open_mfdataset(instant_files, combine="by_coords")
ds_acc = xr.open_mfdataset(accum_files, combine="by_coords")
```

```python
df_inst = ds_inst.to_dataframe().reset_index()
df_acc = ds_acc.to_dataframe().reset_index()

df = df_inst.merge(
    df_acc,
    on=["valid_time", "latitude", "longitude"]
)
```

```python
import numpy as np
df["temp"] = df["t2m"] - 273.15
df["wspd"] = np.sqrt(df["u10"]**2 + df["v10"]**2)
df["pres"] = df["sp"] / 100
df["prcp"] = df["tp"] * 1000
```

```python
df_weather = df[["valid_time", "latitude", "longitude", "temp", "wspd", "prcp", "pres"]]

df_weather.head()
df["prcp_hourly"] = df.groupby(["latitude", "longitude"])["tp"].diff() * 1000
```
